## 1. Domain-Specific Area

This project focuses on **fake news detection** as a domain-specific Natural Language Processing task. Fake news can be understood as misleading, fabricated or unreliable news content that is presented in a format that may appear credible to readers. This is an important problem because digital news articles can spread rapidly through online platforms, making it difficult for readers, journalists and content moderators to manually verify every article. Automated text classification can therefore support the identification of unreliable news content by analysing linguistic patterns within article titles and body text.

Fake news detection is suitable for Natural Language Processing because news articles are primarily text-based and may contain linguistic features that distinguish fake articles from real ones. Recent research describes fake news as a significant digital information problem because false or misleading news can spread widely through modern online platforms and influence public understanding of events [1]. Machine learning and deep learning techniques are therefore commonly investigated as automated approaches for identifying fake news from textual and contextual features [1].

This task is challenging because fake news may be written in a style that imitates genuine news, meaning that simple keyword matching is unlikely to be sufficient for reliable classification. Recent comparative research has shown that fake news classification can be approached using traditional machine learning, convolutional neural networks, transformer-based models and large language models, with each method offering different strengths and limitations [2]. This supports the need to compare more than one modelling approach rather than relying on a single classifier.

A classifier may achieve high accuracy by learning dataset-specific shortcuts rather than genuine indicators of misinformation. This is especially important when fake and real articles come from different sources or collection methods. In such cases, the model may learn source-related patterns instead of learning whether the content itself is reliable. The selected dataset is related to the ISOT fake news dataset structure, where truthful articles are reported as being collected from Reuters, while fake articles are collected from unreliable websites flagged by fact-checking sources [3]. This creates a possible source artefact that must be considered when interpreting model performance.

This domain is appropriate because it allows a clear comparison between statistical and embedding-based text classification models. A traditional statistical model can represent news articles using word-frequency features such as CountVectorizer or TF-IDF, while an embedding-based deep learning model can learn dense word representations and phrase-level patterns. Therefore, fake news detection provides a meaningful classification problem for evaluating whether simpler statistical NLP methods or neural embedding-based models are more effective for binary news classification. At the same time, the project allows critical discussion of whether high model scores reflect genuine fake news detection ability or dataset-specific patterns.

## 4. Evaluation Methodology

The fake news models will be evaluated using supervised binary classification. After preprocessing, the dataset will be split into training and testing sets. The training set fits the models, while the test set evaluates performance on unseen articles and generalisation.

The main evaluation metrics used in this project are accuracy, precision, recall, F1-score and confusion matrix analysis. These metrics are selected because accuracy alone does not show whether the model performs equally well on both fake and real news classes. Recent machine learning evaluation research highlights the importance of selecting appropriate metrics for supervised classification tasks rather than relying on a single performance measure [4]. The formulas used for these metrics are based on true positives (TP), true negatives (TN), false positives (FP) and false negatives (FN) [4], [6].

Accuracy measures the overall proportion of correctly classified articles. It is calculated as:

$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$

Accuracy is useful for understanding general performance, but it may hide class-level errors. For example, a high accuracy score does not necessarily show whether the model is better at detecting fake news or real news.

Precision measures how many articles predicted as a class are actually correct. It is calculated as:

$$
Precision = \frac{TP}{TP + FP}
$$

Precision is important because a fake news detection system should avoid incorrectly labelling real news as fake. A low precision score would mean that the model produces too many false positives.

Recall measures how many actual articles from a class are successfully detected by the model. It is calculated as:

$$
Recall = \frac{TP}{TP + FN}
$$

Recall is important because failing to detect fake news would reduce the usefulness of the classifier. A low recall score would mean that the model misses too many fake news articles.

The F1-score provides a balanced measure between precision and recall. It is calculated as:

$$
F1\text{-}score = 2 \times \frac{Precision \times Recall}{Precision + Recall}
$$

The F1-score is useful when both false positives and false negatives matter. In fake news detection, this is important because classifying fake news as real may allow unreliable content to pass undetected, while classifying real news as fake may reduce trust in legitimate reporting.

A confusion matrix will also be used to examine the number of true positives, true negatives, false positives and false negatives produced by each model. This will allow the evaluation to identify whether a model is more likely to misclassify fake news as real or real news as fake. Such analysis is important because fake news detection is not only about achieving a high overall score, but also about understanding the type of errors made by the classifier.

The models will be compared using the same train-test split and the same evaluation metrics to ensure a fair comparison. The baseline model, statistical model and embedding-based model will be evaluated side by side using a results table and visualisations such as confusion matrices and bar charts. In addition, an ablation test will be used later in the project to examine whether Reuters dateline source artefacts influence model performance. This methodology will make it possible to assess both predictive performance and the reliability of the model comparison.

### 5.10 Cleaning the Text

Text cleaning is an important preprocessing step because raw news articles may contain uppercase letters, punctuation and inconsistent spacing. These elements can increase noise in the dataset and make it harder for the models to learn useful language patterns.

In this project, the **content** column is cleaned by converting text to lowercase, removing punctuation and removing extra spaces. Numbers are retained because they may contain meaningful information in news articles, such as dates, financial figures, statistics or event counts. Keeping numbers helps preserve important contextual information that may be useful for fake news classification.


```python
# create a function to clean text

def clean_text(text):
    # convert the text to lowercase
    text = text.lower()
    
    # remove the punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    # remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text
```

```python
# apply the cleaning function to the content column

news_df["clean_content"] = news_df["content"].apply(clean_text)

# show original and cleaned text for comparison
news_df[["content", "clean_content", "label"]].head()
```

The output shows that the **clean_content** column has been created successfully. The cleaned text has been converted to lowercase, and punctuation has been removed while meaningful numbers are retained. For example, terms such as **9th circuit court** remain understandable after cleaning, which is important because numbers may carry useful information in news articles.

This confirms that the text is now more consistent for model training while still preserving important contextual information. The **clean_content** column will be used as the main text input for both the statistical and embedding-based models.

### 5.13 Train-Test Split

After cleaning the text, the dataset is divided into training and testing sets. The training set is used to train the models, while the testing set is kept separate and used only to evaluate final model performance on unseen data.

This project uses an **80:20 train-test split**, meaning that 80% of the dataset is used for training and 20% is used for testing. This ratio is selected because the dataset is relatively large, containing more than 44000 articles after preprocessing. An 80% training split provides the models with enough examples to learn meaningful fake and real news patterns, while the 20% testing split still leaves a large independent test set for reliable evaluation. A smaller test split, such as 90:10, would provide more training data but less evidence for evaluation, while a larger test split, such as 70:30, would reduce the amount of training data available for the deep learning model. Therefore, 80:20 provides a practical balance between model learning and fair testing.

A fixed random state is used to make the split reproducible. Stratification is also applied so that the proportion of fake and real news articles remains similar in both the training and testing sets. This is important because both classes need to be fairly represented during training and evaluation.


```python
# define input features and target label

X = news_df["clean_content"]
y = news_df["label"]

# split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    # Ensures each split contains the same proportion of labels as the original dataset
    stratify=y  
)

# check the size of each split
# print the shape
print("Training set size:", X_train.shape[0])
print("Testing set size:", X_test.shape[0])
print("\nTraining label distribution:")
print(y_train.value_counts())

print("\nTesting label distribution:")
print(y_test.value_counts())
```

The output shows that the dataset has been split into **35751 training articles** and **8938 testing articles**. This confirms that the 80:20 train-test split has been applied correctly.

The label distribution also shows that stratification has worked as intended. In the training set, there are **18782 fake news articles** and **16969 real news articles**. ## 6. Baseline Performance

The baseline model provides a simple benchmark for comparison with the more advanced models developed later in the project. In this project, the baseline uses CountVectorizer with Multinomial Naive Bayes.

CountVectorizer converts the cleaned article text into a numerical matrix based on word counts. Each article is represented by the frequency of words that appear in the training vocabulary. Multinomial Naive Bayes is then used as the classifier because it is simple, efficient and suitable for text classification tasks that use word-count features.

This baseline is important because it gives a starting performance level for the project. The later TF-IDF Logistic Regression model and embedding-based 1D CNN model should improve on this baseline. If a more complex model performs only slightly better than the baseline, then the additional complexity may not be justified.


```python
# Baseline Model: Countvectorizer with Multinomial Naive Bayes model

# convert the text into word count features
count_vectorizer = CountVectorizer(max_features=5000)

# prepare data for training and testing
X_train_counts = count_vectorizer.fit_transform(X_train)
X_test_counts = count_vectorizer.transform(X_test)

#  create and train the Naive Bayes model
baseline_model = MultinomialNB()
baseline_model.fit(X_train_counts, y_train)

# make the predictions on the test set
baseline_predictions = baseline_model.predict(X_test_counts)
# make sure that baseline model works and trained
# print check to see if the code works
print("Baseline model training completed.")
```

The baseline model was trained using word-count features from the cleaned article text. The Countvectorizer was fitted only on the training data and then applied to the testing data. ## 7. Comparative Classification Methodology
### 7.1 Statistical Model: TF-IDF with Logistic Regression
The first comparative model has TF IDF which goes along with Logistic Regression which stands for "Term Frequency Inverse Document Frequency". It converts text into numerical value by giving greater priority to words that come more frequent in a particular article but are lesser common among the entire dataset.

This process is helpful for detecting fake news as parts may appear more strongly in fake or real news articles. Compared with simple word count features, TF-IDF reduces the influence of very common words and highlights terms that may be more suitable and informative for classification.

Logistic Regression is used as the classifier because it is a strong and interpretable statistical machine learning model for binary classification. In this project, it is used to evaluate if a more refined statistical text representation can improve performance compared with the baseline Countvectorizer and Multinomial Naive Bayes model.

```python
# Statistical Model: TF-IDF and Logistic Regression

# convert text into TF-IDF features
tfidf_vectorizer = TfidfVectorizer(max_features=5000)
# prepare for train and test
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

# create and train the Logistic Regression model
tfidf_lr_model = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)
tfidf_lr_model.fit(X_train_tfidf, y_train)

# predictions on the test set
tfidf_lr_predictions = tfidf_lr_model.predict(X_test_tfidf)
# make sure that TF-IDF Logistic Regression model works and trained
# print check to see if the code works
print("TF-IDF Logistic Regression model training completed.")
```

The TF-IDF vectorizer was initially fit by the help of training data.Then it was applied to the testing data. This totaly reduces data leaking as the model do not learn vocabulary or the importance of word information from the test dataset before evaluating it.

The max_features which equals 5000 sets the feature to limit space to 5000 most priortized terms. These steps makes sure the model is perfoming well while still letting it to learn from a wide range of important words present inside the dataset. Logistic Regression was then effectively trained on TF-IDF features to segregate each article as real or fake.

```python
# evaluate the TF-IDF Logistic Regression model

# accuracy of the TF-IDF Logistic Regression model
tfidf_lr_accuracy = accuracy_score(y_test, tfidf_lr_predictions)
# precision of the TF-IDF Logistic Regression model
tfidf_lr_precision = precision_score(y_test, tfidf_lr_predictions)
# recall of the TF-IDF Logistic Regression model
tfidf_lr_recall = recall_score(y_test, tfidf_lr_predictions)
# f1-score of the TF-IDF Logistic Regression model
tfidf_lr_f1 = f1_score(y_test, tfidf_lr_predictions)

# print the values 
print("TF-IDF Logistic Regression Results")
print("----------------------------------")
print("Accuracy:", round(tfidf_lr_accuracy, 4))
print("Precision:", round(tfidf_lr_precision, 4))
print("Recall:", round(tfidf_lr_recall, 4))
print("F1-score:", round(tfidf_lr_f1, 4))

# print classification report
print("\nClassification Report:")
print(classification_report(y_test, tfidf_lr_predictions, target_names=["Fake", "Real"]))
```

The classification report also shows balanced performance across both classes. The model achieved approximately **0.99** precision, recall and F1 score for both fake and real news articles. ### 7.2 Embedding-Based Model: 1D CNN

The second comparative model is an embedding-based deep learning model using a **1D Convolutional Neural Network**. Unlike CountVectorizer and TF-IDF, which represent text using sparse word-frequency features, the embedding-based model converts words into dense numerical vectors. These vectors are learned during training and allow the model to represent words in a continuous feature space.

A 1D CNN is used because it can learn local word and phrase patterns from text sequences. This is useful for fake news detection because certain short phrases, headline styles or word combinations may help distinguish fake news from real news. The model architecture includes tokenisation, sequence padding, an Embedding layer, a Conv1D layer, a GlobalMaxPooling1D layer, Dropout and Dense layers.

The CNN model is included to evaluate whether an embedding-based neural approach can improve performance compared with the statistical models. Since the TF-IDF Logistic Regression model already achieved very strong results, the CNN will be assessed not only by accuracy but also by whether its added complexity provides meaningful benefits.


```python
# set parameters for the CNN model
# max number of words to keep in the vocabulary
MAX_WORDS = 10000  
# max number of words per article
MAX_LENGTH = 300  
# size of word embedding vectors
EMBEDDING_DIM = 100   
```

The CNN model uses a limited vocabulary of the **10000 most frequent words** to keep training manageable. Each article is padded or shortened to **300 words** so that all input sequences have the same length. The embedding dimension is set to **100**, meaning each word is represented using a 100 value dense vector.


```python
# tokenise the text for the CNN model

# create a tokenizer to convert text into sequences of integers
tokenizer = Tokenizer(num_words=MAX_WORDS, oov_token="<OOV>")
tokenizer.fit_on_texts(X_train)

# convert text into sequences of numbers for train and test
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)

# pad sequences so all articles have the same length
X_train_pad = pad_sequences(X_train_seq, maxlen=MAX_LENGTH, padding="post", truncating="post")
X_test_pad = pad_sequences(X_test_seq, maxlen=MAX_LENGTH, padding="post", truncating="post")

print("Training sequence shape:", X_train_pad.shape)
print("Testing sequence shape:", X_test_pad.shape)
```

The text data has been tokenised and converted into numerical sequences for the CNN model. The tokenizer was fitted only on the training text to avoid data leakage. The same tokenizer was then used to transform the test text.

Padding was applied so that every article has the same sequence length of **300 words**. This is required because neural networks need inputs with consistent dimensions. The output shapes confirm that the training and testing text data have been prepared correctly for the embedding-based model.


```python
# build the 1D CNN model for binary text classification
cnn_model = Sequential()

# input shape (padded text sequences)
cnn_model.add(tf.keras.Input(shape=(MAX_LENGTH,)))

# convert integer-encoded words into dense vector representations
cnn_model.add(Embedding(input_dim=MAX_WORDS, output_dim=EMBEDDING_DIM))

# apply 1D convolution to extract local text features
cnn_model.add(Conv1D(filters=128, kernel_size=5, activation="relu"))

# keep the most important feature from each filter
cnn_model.add(GlobalMaxPooling1D())

# reduce overfitting by dropping 50% during training
cnn_model.add(Dropout(0.5))

# fully connected hidden layer for learning higher level features
cnn_model.add(Dense(64, activation="relu"))

# output layer for binary classification (fake or real)
cnn_model.add(Dense(1, activation="sigmoid"))

# configure the model with loss function, optimiser and evaluation metric
cnn_model.compile(
    loss="binary_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

# display the model architecture and parameter summary
cnn_model.summary()
```

The model summary confirms that the 1D CNN architecture has been built correctly. The Embedding layer outputs sequences with shape **(None, 300, 100)**, meaning each article is represented as a sequence of 300 tokens, and each token is represented using a 100-dimensional embedding vector.

The Conv1D layer then learns local phrase-level patterns from these word embeddings. The GlobalMaxPooling1D layer reduces the sequence output into a fixed-size feature vector by keeping the most important learned features. Dropout is included to reduce overfitting, and the final Dense layer uses a sigmoid activation function for binary classification.

The model contains **1072449 trainable parameters**, which means it is substantially more complex than the baseline and TF-IDF Logistic Regression models. This makes it important to evaluate whether the CNN’s additional complexity leads to better performance.


```python
# train the CNN model

cnn_history = cnn_model.fit(
    X_train_pad,
    y_train,
    epochs=3,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)
```

The CNN model was trained for **3 epochs**. The training accuracy increased from **0.9686** in the first epoch to **0.9994** by the third epoch. The validation accuracy also remained very high, increasing from **0.9973** in the first epoch to **0.9983** in the second and third epochs.

These results show that the CNN was able to learn strong patterns from the tokenised article text. However, the very high training and validation accuracy also suggests that the model may be learning dataset-specific patterns. Therefore, the final test-set evaluation and Reuters removal test are important because they help assess whether the model is learning general fake-news patterns or relying partly on source-related shortcuts.


```python
# make predictions using the CNN model

cnn_probabilities = cnn_model.predict(X_test_pad)
cnn_predictions = (cnn_probabilities > 0.5).astype("int32").flatten()

# calculate evaluation metrics
cnn_accuracy = accuracy_score(y_test, cnn_predictions)
cnn_precision = precision_score(y_test, cnn_predictions)
cnn_recall = recall_score(y_test, cnn_predictions)
cnn_f1 = f1_score(y_test, cnn_predictions)
# print results
print("1D CNN Model Results")
print("--------------------")
print("Accuracy:", round(cnn_accuracy, 4))
print("Precision:", round(cnn_precision, 4))
print("Recall:", round(cnn_recall, 4))
print("F1-score:", round(cnn_f1, 4))
# print classification report
print("\nClassification Report:")
print(classification_report(y_test, cnn_predictions, target_names=["Fake", "Real"]))
```

The 1D CNN model achieved an accuracy of **0.9985**, precision of **0.9986**, recall of **0.9983** and F1-score of **0.9985**. These results show that the embedding-based deep learning model performed extremely well on the fake news classification task.

Compared with the baseline and statistical models, the CNN achieved the highest overall performance. The baseline CountVectorizer with Multinomial Naive Bayes model achieved an accuracy of **0.9503**, while the TF-IDF Logistic Regression model achieved **0.9886**. The CNN improved the accuracy further to **0.9985**, suggesting that the embedding-based model was able to learn highly effective word and phrase-level patterns from the article text.

The classification report also shows balanced performance across both fake and real news classes, with precision, recall and F1-score values close to **1.00** for both classes. This indicates that the CNN model was able to classify both categories consistently, rather than performing well on only one class.

However,the very high performance should be interpreted carefully. Although the CNN performs best on this dataset, the result may partly reflect dataset-specific writing patterns or source-related differences between fake and real articles. Therefore, the model’s real-world performance may be lower when applied to newer news articles from different sources. This limitation is investigated further through the Reuters removal test.

### 7.4 Reuters Removal Test

The model comparison results show very high performance, especially for the TF-IDF Logistic Regression model and the 1D CNN model. However, Section 5 showed that the word **Reuters** appears in **99.82%** of real news articles but only **1.37%** of fake news articles. This means the dataset contains a strong source-related pattern.

To investigate this, a Reuters removal test is carried out. In this test, the word **Reuters** is removed from the cleaned article text, and the TF-IDF Logistic Regression model is trained and evaluated again. The purpose is to check if model performance changes after removing this clear source-related marker.

The TF-IDF Logistic Regression model is used for this test because it is fast, strong and easier to compare against the original statistical result. If performance drops after removing the Reuters marker, this suggests that the original model was partly relying on source-specific information rather than only learning general fake-news language patterns.


```python
# Reuters Dateline Removal Test

# remove the word reuters from the already cleaned text
news_df["clean_content_no_reuters"] = news_df["clean_content"].str.replace(
    r"\breuters\b",
    "",
    regex=True
)

# remove extra spaces after removing Reuters
news_df["clean_content_no_reuters"] = news_df["clean_content_no_reuters"].str.replace(r"\s+"," ",regex=True).str.strip()

# define new input features and target label
X_no_reuters = news_df["clean_content_no_reuters"]
y_no_reuters = news_df["label"]

# use the same 80:20 stratified split
X_train_nr, X_test_nr, y_train_nr, y_test_nr = train_test_split(
    X_no_reuters,
    y_no_reuters,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y_no_reuters
)

# convert text into TF-IDF features
tfidf_vectorizer_nr = TfidfVectorizer(max_features=5000)

X_train_tfidf_nr = tfidf_vectorizer_nr.fit_transform(X_train_nr)
X_test_tfidf_nr = tfidf_vectorizer_nr.transform(X_test_nr)

# train Logistic Regression model
tfidf_lr_model_nr = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)
tfidf_lr_model_nr.fit(X_train_tfidf_nr, y_train_nr)

# make predictions
tfidf_lr_predictions_nr = tfidf_lr_model_nr.predict(X_test_tfidf_nr)

# evaluate performance
nr_accuracy = accuracy_score(y_test_nr, tfidf_lr_predictions_nr)
nr_precision = precision_score(y_test_nr, tfidf_lr_predictions_nr)
nr_recall = recall_score(y_test_nr, tfidf_lr_predictions_nr)
nr_f1 = f1_score(y_test_nr, tfidf_lr_predictions_nr)

print("TF-IDF Logistic Regression After Reuters Removal")
print("-----------------------------------------------")
print("Accuracy:", round(nr_accuracy, 4))
print("Precision:", round(nr_precision, 4))
print("Recall:", round(nr_recall, 4))
print("F1-score:", round(nr_f1, 4))

print("\nClassification Report:")
print(classification_report(y_test_nr, tfidf_lr_predictions_nr, target_names=["Fake", "Real"]))
```

The Reuters removal test creates a new version of the cleaned article text where the word **Reuters** has been removed. Extra spaces are then cleaned again so that the text remains readable and consistent for modelling.

The Reuters-removed text is split into training and testing sets using the same 80:20 stratified split. This keeps the comparison fair because the same model type, split ratio and evaluation metrics are used. The TF-IDF Logistic Regression model is then trained again using the Reuters-removed text, allowing the new result to be compared with the original TF-IDF Logistic Regression result.


```python
# compare original TF-IDF model with Reuters removed TF-IDF model

reuters_test_df = pd.DataFrame({
    "Experiment": [
        "Original TF-IDF + Logistic Regression",
        "Reuters Removed TF-IDF + Logistic Regression"
    ],
    "Accuracy": [
        tfidf_lr_accuracy,
        nr_accuracy
    ],
    "Precision": [
        tfidf_lr_precision,
        nr_precision
    ],
    "Recall": [
        tfidf_lr_recall,
        nr_recall
    ],
    "F1-score": [
        tfidf_lr_f1,
        nr_f1
    ]
})

# round values to 4 decimal places
reuters_test_df = reuters_test_df.round(4)

reuters_test_df
```

## 9. Performance Analysis and Comparative Discussion

In conclusion the three models performed well on the fake news detection task and performance improved as the models became more advanced.  The baseline CountVectorizer with Multinomial Naive Bayes model got an accuracy of 0.9503 and an F1-score of 0.9480. This strong baseline shows that simple characteristics based on word counts can identify useful distinctions between false and real news articles.

The TF-IDF Logistic Regression model got a better accuracy of 0.9886 and an F1-score of 0.9880. We can understand that TF-IDF was better than simple word counts for representing the article text. TF-IDF decreases the influence of common words and gives more priority to words that are more distinctive within specific documents unlike countVectorizer. This helped the Logistic Regression model segregate fake and real news articles more correctly.

Embedding-based 1D CNN model had the best performance, which got an accuracy of 0.9985 and an F1-score of 0.9985. This shows that the CNN learned very effective word and phrase-level patterns from the article text. The CNN also made lesser errors than the other models, showing that embedding-based approach made the strongest result.

The baseline model was the simplest and quickest, but produced the weakest output. The TF-IDF Logistic Regression model provided a strong balance between performance, simplicity and interpretability. It reduced misclassifications alot compared with the baseline while remaining easier to explain than the CNN. The CNN achieved the highest performance, but it also required more preprocessing steps,tokenisation,sequence padding,neural network training and higher computational cost.

The Reuters removal test gave an important finding. Section 5 showed that Reuters appeared in 99.82% of real news articles but only 1.37% of fake news articles. After removing Reuters, the TF-IDF Logistic Regression model’s F1-score decreased from 0.9880 to 0.9819. This shows that the Reuters source marker influenced model performance. However, the model still performed strongly after removal, implying it was not relying entirely on this source pattern and was also learning other language-based features.

Although the CNN achieved the best result, the very high performance should be interpreted carefully. The dataset may contain source-specific, stylistic or topic-related patterns that separate fake and real articles more clearly than in real-world conditions. This means the models may perform extremely well on this dataset but may not generalise equally well to newer articles from different sources. Since this project only uses article text, it also does not include external fact-checking evidence, source credibility information or social media propagation signals.

Overall, the results suggest that embedding-based 1D CNN produced the strongest performance for this experiment. However, the TF-IDF Logistic Regression model remains a highly effective overall and practical alternative because it achieved very high accuracy with lower complexity. Therefore, the choice between the models depends on the intended use case. If maximum predictive performance is the priority, the CNN is the strongest. If interpretability, speed and simplicity are more important, TF-IDF with Logistic Regression may be the better practical choice.


## 10. Project Summary and Reflections

The fake news detection system was created by using NLP techniques including text classification. The aim was to compare traditional statistical model and deep learning model based on embeddings. For this purpose, the dataset was gathered by combining fake and real news articles and preprocessing the text data. 

Three models were developed for the project: baseline, statistical and CNN. The former used Count Vectorizer together with Multinomial Naive Bayes resulting in 0.9503 accuracy. The latter used TF-IDF together with Logistic Regression and showed 0.9886 accuracy. The 1D CNN based on embedding provided 0.9985 accuracy and 0.9985 F1-Score. Thus, it can be stated that all three approaches were efficient but CNN model performed the best. 

The project proves the significance of good text representation in the context of text classification task. Word count-based features could be used successfully but TF-IDF gave the more meaningful representation by highlighting unique terms. CNN proved the best performance due to the dense representation of words and phrases. However, TF-IDF Logistic Regression also worked pretty well. 

One of the strengths of the work is that the project was not limited to reporting high scores of the models used. For example, the Reuters source-pattern check and Reuters removal test investigated whether there was a source-based feature that would make the dataset easier for separation. The TF-IDF Logistic Regression F1-score decreased from 0.9880 to 0.9819 when Reuters were removed from the training set.

This dataset can have source-specific or stylistic features that will help distinguish between fake and true news.
